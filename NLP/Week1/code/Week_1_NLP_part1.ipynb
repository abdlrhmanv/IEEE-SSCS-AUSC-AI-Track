# **Stemming**

![what_is_stemming?](1709744731810.png)

```python
'''
stemming is :
Technique that reduce word to its base or root form by removing prefixes and suffixes
the root form or stem obtained through stemming may not always be a valid word but serves as a
representation for words with similar meanings in computational tasks
'''

# Example with common prefiexes and suffixes
# [ 1 ] unhappy = un + happy
# [ 2 ] replay = re + play
# [ 3 ] incomplete = in + complete
# [ 4 ] disagree = dis + agree
# [ 5 ] running = runn + ing 
# [ 6 ] played = play + ed
# [ 7 ] production = product + ion
```

## **Types of Stemming**

![types_of_stemming](popular_stemming_algorithms.webp)

```python
from nltk import PorterStemmer , SnowballStemmer , LancasterStemmer

text = "Production is the most important phase in software Engineering life cycle"

porterstemmer = PorterStemmer()               # remove common suffixes from english words, and prefiexes
snowballstemmer = SnowballStemmer("english")  # supports multiple languages, making it a multilingual stemmer
lancasterstemmer = LancasterStemmer()         # known for being more aggressive and faster than other stemmers

tokens = word_tokenize(text)

porter_tokens = []
snowball_tokens = []
lancaster_tokens = []
for i in tokens :
    porter_tokens.append(porterstemmer.stem(i))
    snowball_tokens.append(snowballstemmer.stem(i))
    lancaster_tokens.append(lancasterstemmer.stem(i))

print(porter_tokens)
print(snowball_tokens)
print(lancaster_tokens)
```

```python
from nltk import ISRIStemmer 

text = "ضاقت فلما استحكمت حلقاتها فرجت و كنت أظنها لا تفرج"

tokens = word_tokenize(text)

stemmer = ISRIStemmer()

new_tokens = [ stemmer.stem(i) for i in tokens]

print(tokens)
print(new_tokens)
```

# **Lemmatization**

![lemmatization](lemmatization.png)

### stemming can lead to over stemming or under stemming, where words can be
### reduced to non existent roots or to many forms are grouped together
### but Lemmatization reduces words to their base form known as a lemma

### unlike stemming, it considers the word's meaning and part of speech ( pos ),
### and ensures that the base form is a valid word, this makes lemmatization more accurate
### as it avoids generating non dictionary words.
### **it is slower than stemming but better**

## **Lemmatization without POS**

```python
import nltk

nltk.download("wordnet")
```

```python
from nltk import WordNetLemmatizer

text = "Production is the most important phase in AI and Machine Learning"

tokens = word_tokenize(text)

lemmatizer = WordNetLemmatizer()

new_tokens = [ lemmatizer.lemmatize(i) for i in tokens]

print(tokens)
print(new_tokens)
```

## **Lemmatization using pos**

```python
from nltk import WordNetLemmatizer , pos_tag , word_tokenize

from nltk.corpus import wordnet

def get_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('R'):
        return wordnet.ADV
    elif tag.startswith('N'):
        return wordnet.NOUN
    return wordnet.NOUN

lemmatizer = WordNetLemmatizer()
sentence = "omar is running faster than the other students"
tokens = word_tokenize(sentence)
tags = pos_tag(tokens)

new_tokens = []
for word,tag in tags :
    lemma = lemmatizer.lemmatize(word,get_pos(tag))
    new_tokens.append(lemma)

print(new_tokens)
```

```python
from nltk import WordNetLemmatizer , pos_tag , word_tokenize
from nltk.corpus import wordnet

def get_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('R'):
        return wordnet.ADV
    elif tag.startswith('N'):
        return wordnet.NOUN
    return wordnet.NOUN

sentence = "omar is running faster than the other students"
tokens = word_tokenize(sentence)
tags = pos_tag(tokens)

new_tokens = []
for word,tag in tags :
    lemmatized_Word = lemmatizer.lemmatize(word,get_pos(tag))
    new_tokens.append(lemmatized_Word)

print(new_tokens)

```

# **Named Entity Recognition**
### **Key technique in NLP that involves identifying and classifying named**
### **entites in text into predifined categories**
![NER](NER.png)

## --------------------
## **Methods of NER**
## --------------------
### **Lexicon Based Method**
#### This method uses a dictionay of known entity names
#### This process involves checking if any of these words are present in a given text, However
#### This approach isn't commonly used because it requires constant updating and careful maintenance
#### of the dictionary to stay accurate and effective
## --------------------
### **Rule Based Method**
## --------------------
#### it uses a set of predined rules which helps in extraction of information,
#### These rules are based on patterns and context
## --------------------
### **Machine Learning Based Method**
## --------------------
#### we use multi class classificatin, Train model on labeled examples where each entity is categorized,
#### in addition to labeling model also required a deep understanding of context which makes it 
#### a challenging task for simple machine learning algorithms
## --------------------
### **Deep Learning Based Method**
## --------------------
#### word Embeddings : capture meaning of words in context
#### Automatic Learning : Deep models learn complex patterns without manual feature engineering
#### Higher Accuracy : performs well on large varied datasets
